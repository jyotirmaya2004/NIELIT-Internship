## Machine Learning

Machine Learning is a branch of Artificial Intelligence in which a computer learns patterns from data and uses those patterns to make predictions, decisions, or recommendations without being explicitly programmed for every rule.

> "A computer program is said to learn from experience **E** with respect to some class of tasks **T** and performance measure **P**, if its performance at tasks in **T**, as measured by **P**, improves with experience **E**."

###### Tom M. Mitchell, 1997

#### Understanding T, P, and E in ML

- **T - Task:** The work the machine has to perform.
- **P - Performance measure:** The way we judge how well the machine is performing.
- **E - Experience:** The data or examples from which the machine learns.

Example: Email spam detection

- **Task:** Classify emails as spam or not spam.
- **Performance measure:** Accuracy, precision, recall, or F1-score.
- **Experience:** A collection of old emails already labelled as spam or not spam.

Why T, P, and E are important:

1. They give clarity to the learning objective.
2. They help measure whether the model is improving.
3. They identify the learning source, usually data.
4. They make model evaluation scientific and measurable.

#### How ML Is Different from Traditional Programming

In traditional programming, humans write rules and the computer follows those rules to produce output.

**Traditional programming:**

`Input + Rules -> Output`

Example: A programmer writes fixed rules to calculate simple interest.

In machine learning, we give the computer data and the expected output. The machine learns the rules or patterns by itself.

**Machine learning:**

`Input + Output -> Learned Model/Rules`

Example: A spam filter learns from many labelled emails instead of using only manually written rules.

#### When Do We Need ML?

We use machine learning when:

1. Rules are hard to define manually, such as face recognition or voice recognition.
2. Data is large and complex, such as customer behaviour or sensor data.
3. The system must learn and improve over time, such as a recommendation system.
4. Hidden patterns exist in the data, such as fraud detection.
5. We want automation or prediction, such as weather prediction or price prediction.
6. The environment changes frequently, such as online advertisements or stock market analysis.

#### Types of Machine Learning

Machine learning is commonly divided into three main types:

1. **Supervised Learning**
2. **Unsupervised Learning**
3. **Reinforcement Learning**

### 1. Supervised Learning

Supervised learning is a type of ML where the model learns from labelled data. Labelled data means each input already has the correct output.

Examples:

- Predicting house price from area, location, and number of rooms.
- Classifying an email as spam or not spam.
- Predicting whether a student will pass or fail.

Main tasks in supervised learning:

1. **Regression:** Predicts a continuous numeric value.
   - Example: house price, temperature, salary, sales amount.
2. **Classification:** Predicts a category or class label.
   - Example: spam/not spam, pass/fail, disease/no disease.

Common algorithms:

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine
- K-Nearest Neighbors
- Naive Bayes

### 2. Unsupervised Learning

Unsupervised learning is a type of ML where the model learns from unlabelled data. The model tries to find hidden structure, groups, or patterns on its own.

Examples:

- Grouping customers based on buying behaviour.
- Finding similar documents.
- Detecting unusual transactions.

Main tasks in unsupervised learning:

1. **Clustering:** Groups similar data points together.
   - Example: customer segmentation.
2. **Association:** Finds relationships between items.
   - Example: customers who buy bread may also buy butter.
3. **Dimensionality Reduction:** Reduces the number of features while keeping important information.
   - Example: PCA for simplifying large datasets.

Common algorithms:

- K-Means Clustering
- Hierarchical Clustering
- DBSCAN
- Apriori Algorithm
- Principal Component Analysis

### 3. Reinforcement Learning

Reinforcement learning is a type of ML where an agent learns by interacting with an environment. The agent takes actions and receives rewards or penalties. The goal is to learn actions that maximize total reward.

Important terms:

- **Agent:** The learner or decision maker.
- **Environment:** The world in which the agent acts.
- **Action:** A choice made by the agent.
- **Reward:** Feedback received after an action.
- **Policy:** The strategy used by the agent to choose actions.

Examples:

- Game playing systems.
- Robot movement.
- Self-driving cars.
- Dynamic pricing systems.

#### Basic ML Workflow

A typical machine learning project follows these steps:

1. **Define the problem:** Decide what prediction or decision is needed.
2. **Collect data:** Gather relevant data from files, databases, sensors, websites, or surveys.
3. **Prepare data:** Clean missing values, remove duplicates, fix errors, and convert data into usable form.
4. **Explore data:** Understand patterns using statistics and visualizations.
5. **Select features:** Choose the important input variables.
6. **Split data:** Divide data into training and testing sets.
7. **Train the model:** Let the algorithm learn from training data.
8. **Evaluate the model:** Check performance using suitable metrics.
9. **Improve the model:** Tune parameters, add better features, or try another algorithm.
10. **Deploy the model:** Use the model in a real application.

#### Important Terms in ML

- **Dataset:** Collection of data used for learning.
- **Feature:** Input variable used for prediction.
- **Label/Target:** Correct output value in supervised learning.
- **Model:** The learned mathematical pattern or rule.
- **Training data:** Data used to teach the model.
- **Testing data:** Data used to check model performance on unseen examples.
- **Overfitting:** Model performs very well on training data but poorly on new data.
- **Underfitting:** Model is too simple and performs poorly on both training and testing data.
- **Generalization:** Ability of a model to perform well on new, unseen data.

#### Evaluation Metrics

For classification problems:

- **Accuracy:** Percentage of correct predictions.
- **Precision:** Out of predicted positive cases, how many are actually positive.
- **Recall:** Out of actual positive cases, how many were correctly predicted.
- **F1-score:** Balance between precision and recall.
- **Confusion Matrix:** Table showing correct and incorrect classifications.

For regression problems:

- **MAE:** Mean Absolute Error.
- **MSE:** Mean Squared Error.
- **RMSE:** Root Mean Squared Error.
- **R-squared:** Shows how well the model explains variation in the target variable.

#### Advantages of Machine Learning

1. Handles large and complex data.
2. Finds patterns that humans may miss.
3. Improves performance with more experience.
4. Helps automate repetitive decision-making tasks.
5. Supports prediction, classification, recommendation, and anomaly detection.

#### Limitations of Machine Learning

1. Requires good quality data.
2. Bad or biased data can produce bad or biased predictions.
3. Some models are difficult to interpret.
4. Training can require time and computing power.
5. ML models may fail when future data is very different from training data.

#### Summary

Machine Learning allows computers to learn from data and improve their performance over time. It is useful when manual rules are difficult, data is large, or predictions are required. The three major types are supervised learning, unsupervised learning, and reinforcement learning. A good ML system needs a clear task, useful experience, a proper performance measure, clean data, and careful evaluation.
