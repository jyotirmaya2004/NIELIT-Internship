### Here the Age columns in normally distributed that means we have to use empricial formula to calculate the Boundaries

```python
upper_boundary=df['Age'].mean()+3*df['Age'].std()
lower_boundary=df['Age'].mean()-3*df['Age'].std()
print('Lower boundary: ',lower_boundary)
print('Upper boundary: ',upper_boundary)
```

```python
# Boxplot with boundaries
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Age'], color='lightblue')
plt.axvline(x=lower_boundary, color='red', linestyle='--', label='Lower Bound (±3σ)')
plt.axvline(x=upper_boundary, color='red', linestyle='--', label='Upper Bound (±3σ)')
plt.title('Box Plot - Age with ±3σ Boundaries')
plt.xlabel('Age')
plt.legend()
plt.show()
```
