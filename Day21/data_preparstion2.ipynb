```python
from sklearn.impute import SimpleImputer


# Impute Missing Values with Mode (for categorical columns)
df_impute = df.copy()
mode_imputer = SimpleImputer(strategy='most_frequent')
df_impute['location'] = mode_imputer.fit_transform(df_impute[['location']])
print(df['location'].isnull().sum().sum())
print(df_impute['location'].isnull().sum().sum())
```

```python
#Forward Fill (uses previous value)

df_ffill = df.copy()
#df_ffill.fillna(method='ffill', inplace=True)
df_ffill.ffill(inplace=True)

print(df_ffill.isnull().sum().sum())
print(df.isnull().sum().sum())
```

```python
#Backward Fill (uses next value)
df_bfill = df.copy()
#df_bfill.fillna(method='bfill', inplace=True)

df_bfill.bfill(inplace=True)

print(df_bfill.isnull().sum())
print(df.isnull().sum().sum())
```
