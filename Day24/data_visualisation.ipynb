```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#check all sns built in datasets
print(sns.get_dataset_names())
```

```python
df=sns.load_dataset('iris')
tips=sns.load_dataset('tips')
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

#### Bar Graph

```python
#bar plot
plt.figure(figsize=(10,5))
sns.barplot(x='species',y='sepal_length',data=df,estimator=np.mean,hue='species',palette='Set2', saturation=1)
plt.title('Average Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Sepal Length')
plt.show()
```

```python
#bar plot for tips dataset
plt.figure(figsize=(10,5))
sns.barplot(x='day',y='total_bill',data=tips,estimator=np.mean,hue='day',palette='Set3', saturation=1)
plt.title('Average Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.show()
```

#### Count Plot

```python
plt.figure(figsize=(10,5))
sns.countplot(x='species',data=df,hue='species',palette='Set1')
plt.title('Count of Each Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
#count plot for tips dataset
plt.figure(figsize=(10,5))
sns.countplot(x='day',data=tips,hue='day',palette='Set2')
plt.title('Count of Each Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

#### Pie Chart

```python
species_counts = df['species'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('Set1'))
plt.title('Distribution of Iris Species')
plt.axis('equal')
plt.show()
```

```python
#pie plot for tips dataset
plt.figure(figsize=(8,6))
day_counts = tips['day'].value_counts()
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('Set2'),wedgeprops={"edgecolor": "black"})
plt.title('Distribution of Days in Tips Dataset')
plt.axis("equal")
plt.show()
```

## Univariate plot for numerical data

#### Histogram

```python
plt.figure(figsize=(10,5))
plt.hist(df['sepal_length'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
#histogram for tips dataset
plt.figure(figsize=(10,5))
sns.histplot(tips['total_bill'], bins=20, kde=True, color='lightgreen')
plt.title('Distribution of Total Bill with KDE')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

#### Density plot

```python
plt.figure(figsize=(10,5))
sns.kdeplot(df['sepal_length'], fill=True, color='purple')
plt.title('KDE Plot of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Density')
plt.show()
```

#### Boxplot

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='species',y='sepal_length',data=df,hue='species',palette='Set3')
plt.title('Boxen Plot of Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Length')
plt.show()
```

## Bivariate Plot

#### Scatter plot

```python
#scatter plot
plt.figure(figsize=(10,5))
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue='species',palette='Set1')
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
#scatter plot for tips dataset
plt.figure(figsize=(10,5))
sns.scatterplot(x='total_bill',y='tip',data=tips,hue='day',palette='Set1',s=100,edgecolor='black')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

#### Line Plot

#### Bar Graph

```python
#barplot showing mean sepal width by species
plt.figure(figsize=(10,5))
sns.barplot(x='species',y='sepal_width',data=df,estimator=np.mean,hue='species',palette='Set2', saturation=1)
plt.title('Average Sepal Width by Species')
plt.xlabel('Species')
plt.ylabel('Average Sepal Width')
plt.show()
```

#### Violin Plot

```python
#violin plot for tips dataset showing distribution of total bill by day
plt.figure(figsize=(10,5))
sns.violinplot(x='day',y='total_bill',data=tips,hue='day',palette='Set2')
plt.title('Violin Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

#### Line Plot

```python
#Line plot of petal width for each species over index
sns.lineplot(x=df.index,y='petal_width',data=df,hue='species',palette='Set1',marker='o',markersize=5,linewidth=2)
plt.title('Line Plot of Petal Width over Index')
plt.xlabel('Index')
plt.ylabel('Petal Width')
plt.show()
```

#### Contigency Table

```python
#Add a new categorical column to tips dataset based on total bill amount
def assign_flower_type(species):
	if species == 'setosa':
		return 'Type A'
	elif species == 'versicolor':
		return 'Type B'
	else:
		return 'Type C'
df['flower_type'] = df['species'].apply(assign_flower_type)
df.head()

#contingency table of flower type vs species
contingency_table = pd.crosstab(df['flower_type'], df['species'])
print(contingency_table)
```

#### Heat Map

```python
#heatmap for correlation matrix of iris dataset
plt.figure(figsize=(8,6))
correlation_matrix = df.drop(columns=['species','flower_type']).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Iris Dataset')
plt.show()
```

```python
#heatmap for tips dataset showing correlation between total bill, tip, and size
plt.figure(figsize=(8,6))
correlation_matrix_tips = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(correlation_matrix_tips, annot=True, cmap='YlGnBu', linewidths=0.5)
plt.title('Correlation Matrix of Tips Dataset')
plt.show()
```
