## Data preparation

##### Data preparation or preprocessing is the process of cleaning, transforming, and organizing raw data into a structured and unusable format before it is analysed.

### Why is data preparation important?
1. improve data quality
2. handles missing values
3. increase model accuracy
4. ensures efficiency
5. enables vette insights
6. supports feature engineering
7. reduces overfitting
8. ensures data compability
9. facilitates reproducibility

#### Data preparation steps
1. data collection
2. data cleaning
3. data transformation
4. data reduction
5. data splitting

###### Data scientists spends 80% of time for data preparation

Kay task in data cleaning


1. handling missing data
2. outlier detection and handling
3. data scaling and transformation
4. encoding categorical variables
5. handling duplicates
6. dealing with inconsitent data

```python
#categorical columns
cat_col=[col for col in df.columns if df[col].dtype=="object"]
print("Categorical data",cat_col)

#numerical columns
num_col=[col for col in df.columns if df[col].dtype!="object"]
print("Numerical data : ",num_col)
```

```python
#categorical columns
cat_col1=df.select_dtypes(include='object').columns
num_col1=df.select_dtypes(include='number').columns

print("categorical columns : ",cat_col1.tolist())

print("Numerical columns : ",num_col1.tolist())
```
