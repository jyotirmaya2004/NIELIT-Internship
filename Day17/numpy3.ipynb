```python
import numpy as np
arr = np.array([1, 2])
scalar=5

#broadcasting
result=arr+scalar
print(result)

matrix = np.array([[1, 2], [3, 4]])
print(matrix+arr)
```

#### random number generation

```python
np.random.seed(0)
random_array = np.random.rand(3)
random_array2=np.random.randint(1,10,5) # Generate 5 random integers between 1 and 10
random_normal=np.random.normal(0,1,4) #np.random.normal(mean, std_dev, size)
print("random_array : ",random_array)
print("random_array2 : ",random_array2)
print("random_normal : ",random_normal)
```

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])

print(np.dot(A,B))
print(np.matmul(A,B))
print(np.linalg.inv(A))
print(np.linalg.det(A)) #determinant of A
```
